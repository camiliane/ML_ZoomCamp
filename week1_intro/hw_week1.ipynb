```python
#Q6 - median value of horsepower
median_before = df.horsepower.median()
counts = df.horsepower.value_counts()
df_after = df.copy()
df_after.horsepower = df_after.horsepower.fillna(counts.index[0])
median_after = df_after.horsepower.median()
print(median_before, median_after)
#yes, it decreased
```

```python
#Q7. Sum of weights
cars_from_asia = df[df.origin=='Asia'][['vehicle_weight', 'model_year']]
X = cars_from_asia.head(n=7).values
#X_T = X.T
XTX = np.matmul(X.T, X)
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
XTX_inv = np.linalg.inv(XTX)
w1 = np.matmul(XTX_inv, X.T)
w = np.matmul(w1, y)
print(sum(w))
```
